## 1. Flujo del juego

**1. Crear los 4 tableros** (cada uno es una matriz 10x10 de `_`)
- `t_barcos_rival`: donde la máquina esconde sus barcos (secreto)
- `t_disparos_rival`: los disparos de la máquina contra mí
- `t_barcos_jugador`: mis barcos
- `t_disparos_jugador`: mis disparos contra la máquina

**2. Crear las flotas** con `crear_flota()`
- `flota_rival` y `flota_jugador`
- 6 barcos cada una: uno de eslora 4, dos de eslora 3 y tres de eslora 2
- Sin salirse del tablero y sin solaparse

**3. Colocar los barcos** con `colocar_barcos()`
- La flota del rival va en `t_barcos_rival`
- La flota del jugador va en `t_barcos_jugador`

**4. Jugar por turnos** con `turnos()`
- Empieza el jugador
- Si aciertas (`X`), repites turno. Si fallas (`A`), dispara la máquina
- La máquina dispara al azar y nunca repite casilla
- Antes de cada disparo se enseñan mis barcos y mis disparos con `pintar()`

**5. Fin de la partida**
- Gana quien deja al otro sin ninguna `O` en su tablero de barcos

**Símbolos:** `_` mar · `O` barco · `X` tocado · `A` agua
